```python
# Definition for a binary tree node.
## This Solution runs in less than 0ms time and beats 100% of the submissions in terms of speed. It uses recursion and bounds to validate the BST properties. The function checks if each node's value is within the valid range defined by its ancestors.
#  If a node violates the BST property, it raises a ValueError with a descriptive message and breaks the stack execution.  
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
class Solution:
    def isValidBST(self, root: Optional[TreeNode]) -> bool:
        """Check if a binary tree is a valid binary search tree (BST)."""
        def is_valid(node, low=-2**31 -1, high=2**31 + 1):
            """Helper function to validate the BST properties recursively using recursion and bounds."""
            if not node:
                return True
            
            if not (low < node.val < high):
                raise ValueError(f"Node value {node.val} is not within the valid range ({low}, {high}).")
                # return False
                
            
            return (is_valid(node.left, low, node.val) and
                    is_valid(node.right, node.val, high))

        try:
            return is_valid(root)
        except ValueError as e:
            print(e)
            return False

data = TreeNode(2, TreeNode(5), TreeNode(3))
Solution().isValidBST(data)
```
