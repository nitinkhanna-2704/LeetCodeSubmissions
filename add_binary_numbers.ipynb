```python

class Solution:
    def addBinary(self, a: str, b: str) -> str:
        """Add two binary strings and return their sum as a binary string."""
        bit_map = {
            '0' : {'0':'0',
                    '1' : '1'
            },   '1' : {'0':'1',
                    '1' : '0'
            }
        }

        longer_num = a if len(a) > len(b) else b
        shorter_num = b if len(a) > len(b) else a
        carry = '0'
        new_num = ''

        def addbits(bit1, bit2, carry):
            """Add two binary bits along with a carry. This is using base 2 property of binary number"""
            total = int(bit1) + int(bit2) + int(carry)
            new_carry = '1' if total >= 2 else '0'
            bit_to_add = str(total % 2)
            return bit_to_add, new_carry
        

        for i in range(1, len(longer_num) + 1):
            lv = longer_num[-i]
            shv = shorter_num[-i] if i <= len(shorter_num) else '0'
            bit_to_add, carry = addbits(lv, shv, carry)
            new_num = bit_to_add + new_num
        

        if carry == '1':
            new_num = carry + new_num
        print( new_num)
        return new_num

# Solution().addBinary('11','1')
```
