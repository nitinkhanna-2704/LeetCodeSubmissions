## Wordsearch 

### Should be solved using Backtracking algorithm. 

Given an m x n grid of characters board and a string word, return true if word exists in the grid.

The word can be constructed from letters of sequentially adjacent cells, where adjacent cells are horizontally or vertically neighboring. The same letter cell may not be used more than once.


Constraints:

<li> m == board.length
<li> n = board[i].length
<li> 1 <= m, n <= 6
<li> 1 <= word.length <= 15

board and word consists of only lowercase and uppercase English letters.

 

```python
word='1212'
list(word).pop()
```

```python
class Solution:
    def exist(self, board: List[List[str]], word: str) -> bool:
        M = len(board) #M X N Board
        N = len(board[0])   

        def dfs(row,col,word):
            if not word:
                return True
            
            if row < 0 or row >=M or \
               col < 0 or col >=N or \
               board[row][col] !=word[0]:
                print(f"Backtracking from row {row}, col {col} for word {word}")
                return False
            
            directions = [(0,1),(0,-1),(1,0),(-1,0)]
            temp = board[row][col]
            board[row][col] = '$'
            for dr, dc in directions:
                if dfs(row+dr, col+dc, word[1:]):
                    print(f"Found word {word} at row {row}, col {col}")
                    return True
            board[row][col] = temp
            return False

        for i in range(M):
            for j in range(N):
                if dfs(i,j,word):
                    return True

        return False 
check = Solution()
check.exist([["A","B","C","E"],["S","F","C","S"],["A","D","E","E"]], "SEE")
```
