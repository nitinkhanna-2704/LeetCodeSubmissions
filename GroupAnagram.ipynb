Given an array of strings strs, group the anagrams together. You can return the answer in any order.

 

Example 1:

Input: strs = ["eat","tea","tan","ate","nat","bat"]

Output: [["bat"],["nat","tan"],["ate","eat","tea"]]

Explanation:

There is no string in strs that can be rearranged to form "bat".
The strings "nat" and "tan" are anagrams as they can be rearranged to form each other.
The strings "ate", "eat", and "tea" are anagrams as they can be rearranged to form each other.
Example 2:

Input: strs = [""]

Output: [[""]]

Example 3:

Input: strs = ["a"]

Output: [["a"]]

 

Constraints:

<li> 1 <= strs.length <= 104
<li> 0 <= strs[i].length <= 100
<li> strs[i] consists of lowercase English letters.

```python
class Solution:
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        """
        Groups anagrams from a list of strings.
        Returns a list of lists, where each sublist contains words that are anagrams of each other.
        """
        anagrams = {}
        for word in strs:
            
            sorted_word = ''.join(sorted(word))
            if sorted_word not in anagrams:
                #if the sorted word is not in the dictionary, create a new entry with the current word as the first element of the list
                anagrams[sorted_word] = [word]
            else:
                #append the current word to the existing list of anagrams for the sorted word
                anagrams[sorted_word].append(word)
            ##DEBUGGING: print the current state of the anagrams dictionary after processing each word
            print(anagrams)
        
        return list(anagrams.values())

group_anagrams = Solution().groupAnagrams(["c","c"])
group_anagrams
        
```
