```python
class Solution:
    def combine(self, n: int, k: int) -> List[List[int]]:
        """ function to find unique combinations in which 1-N can be arranged with k places """
        res = []
        comb = []

        def backtrack(start):
            """ Recursive Function to backtrack the list"""
            # print(comb,start)
            if len(comb) == k:
                res.append(comb[:])
                return
            
            for num in range(start, n + 1):
                comb.append(num)
                backtrack(num + 1)
                # print(comb)
                comb.pop()

        backtrack(1)
        return res
    

Solution().combine(6,4)
```
