### Swap Nodes in Pairs

Given a linked list, swap every two adjacent nodes and return its head. You must solve the problem without modifying the values in the list's nodes (i.e., only nodes themselves may be changed.)

 

```python
# Definition for singly-linked list.
class ListNode:
    """ 
    Class to make node objects
    """
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution:
    def swapPairs(self, head: Optional[ListNode]) -> Optional[ListNode]:
        """
        This is a classic 2 pointer problem. Every odd node N changes with N+1 if N+1 is not None
        """
        

        if head is None or head.next is None:
            return head

        odd_ptr = head
        even_ptr = head.next

        while even_ptr and odd_ptr:
           even_ptr.val , odd_ptr.val = odd_ptr.val , even_ptr.val 
           odd_ptr = even_ptr.next 
           even_ptr = odd_ptr.next if odd_ptr else None

        print(head)

        return head


result = Solution().swapPairs(ListNode(1,ListNode(2,ListNode(3,ListNode(4,None)))))
result
```
